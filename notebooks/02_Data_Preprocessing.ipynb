```python
def preprocess_chunk(chunk):
    # Clean column names
    chunk.columns = [
        column.strip()
        for column in chunk.columns
    ]

    # Clean labels
    chunk["Label"] = (
        chunk["Label"]
        .apply(clean_label)
        .replace(LABEL_MAPPING)
    )

    # Convert infinite values to NaN
    chunk[numeric_features] = (
        chunk[numeric_features]
        .replace([np.inf, -np.inf], np.nan)
    )

    # Remove rows with missing feature values
    chunk = chunk.dropna(
        subset=numeric_features
    )

    # Remove duplicate rows within this chunk
    chunk = chunk.drop_duplicates()

    # Keep only the required columns
    chunk = chunk[
        numeric_features + ["Label"]
    ]

    return chunk


print("Full-dataset preprocessing function created successfully.")
```

```python
import hashlib

def row_hash(row):
    row_string = "||".join(
        str(value) for value in row
    )
    return hashlib.md5(
        row_string.encode("utf-8")
    ).hexdigest()


print("FINAL DUPLICATE-DETECTION FUNCTION CREATED")
print("=" * 80)
print("Method: Row-level MD5 hashing")
print("Purpose: Detect duplicates across chunk boundaries")
```

```python
# Notebook 02 - Cell 52

test_file = csv_files[0]

final_test_output = os.path.join(
    FULL_PROCESSED_DIR,
    "final_cleaned_" + os.path.basename(test_file)
)

seen_hashes = set()

input_rows = 0
valid_rows = 0
duplicate_rows = 0
chunk_number = 0

if os.path.exists(final_test_output):
    os.remove(final_test_output)

print("CROSS-CHUNK DUPLICATE REMOVAL TEST")
print("=" * 80)
print("Input :", os.path.basename(test_file))
print("Output:", os.path.basename(final_test_output))

for chunk in pd.read_csv(
    test_file,
    chunksize=CHUNK_SIZE,
    low_memory=False
):
    chunk_number += 1
    input_rows += len(chunk)

    cleaned_chunk = preprocess_chunk(chunk)

    if len(cleaned_chunk) == 0:
        continue

    row_hashes = cleaned_chunk.apply(
        row_hash,
        axis=1
    )

    keep_mask = ~row_hashes.isin(seen_hashes)

    duplicate_rows += int((~keep_mask).sum())

    new_hashes = row_hashes[keep_mask]
    seen_hashes.update(new_hashes)

    unique_chunk = cleaned_chunk.loc[
        keep_mask
    ]

    valid_rows += len(unique_chunk)

    unique_chunk.to_csv(
        final_test_output,
        mode="a",
        header=not os.path.exists(final_test_output),
        index=False
    )

    print(
        f"Chunk {chunk_number}: "
        f"input={len(chunk):,}, "
        f"unique_output={len(unique_chunk):,}, "
        f"duplicates_removed={int((~keep_mask).sum()):,}"
    )

print("\n" + "=" * 80)
print("CROSS-CHUNK TEST COMPLETE")
print("Input rows:", f"{input_rows:,}")
print("Final rows:", f"{valid_rows:,}")
print("Duplicates removed:", f"{duplicate_rows:,}")
print("Output:", final_test_output)
```

```python
# Notebook 02 - Cell 54

FINAL_PROCESSED_DIR = os.path.join(
    PROCESSED_DIR,
    "CIC-IDS2017_final"
)

os.makedirs(
    FINAL_PROCESSED_DIR,
    exist_ok=True
)

final_file_stats = []

print("=" * 90)
print("        FINAL CIC-IDS2017 PREPROCESSING")
print("=" * 90)

for file in csv_files:

    file_name = os.path.basename(file)

    final_output = os.path.join(
        FINAL_PROCESSED_DIR,
        "final_" + file_name
    )

    if os.path.exists(final_output):
        os.remove(final_output)

    seen_hashes = set()

    input_rows = 0
    rows_after_cleaning = 0
    duplicates_removed = 0
    chunk_number = 0

    print("\n" + "=" * 90)
    print("Processing:", file_name)
    print("=" * 90)

    for chunk in pd.read_csv(
        file,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ):

        chunk_number += 1
        input_rows += len(chunk)

        # Basic preprocessing
        cleaned_chunk = preprocess_chunk(chunk)

        if cleaned_chunk.empty:
            continue

        # Generate row hashes
        row_hashes = cleaned_chunk.apply(
            row_hash,
            axis=1
        )

        # Keep only first occurrence
        keep_mask = ~row_hashes.isin(
            seen_hashes
        )

        duplicate_count = int(
            (~keep_mask).sum()
        )

        duplicates_removed += duplicate_count

        # Register newly seen rows
        seen_hashes.update(
            row_hashes[keep_mask]
        )

        unique_chunk = cleaned_chunk.loc[
            keep_mask
        ]

        rows_after_cleaning += len(
            unique_chunk
        )

        # Save final chunk
        unique_chunk.to_csv(
            final_output,
            mode="a",
            header=not os.path.exists(final_output),
            index=False
        )

        print(
            f"Chunk {chunk_number}: "
            f"input={len(chunk):,}, "
            f"output={len(unique_chunk):,}, "
            f"duplicates={duplicate_count:,}"
        )

    final_file_stats.append({
        "File": file_name,
        "Input_Rows": input_rows,
        "Final_Rows": rows_after_cleaning,
        "Duplicates_Removed": duplicates_removed,
        "Total_Removed": input_rows - rows_after_cleaning
    })

    print(
        f"\nCompleted: {file_name}"
    )

    print(
        f"Input rows: {input_rows:,}"
    )

    print(
        f"Final rows: {rows_after_cleaning:,}"
    )

    print(
        f"Duplicates removed: {duplicates_removed:,}"
    )

    print(
        f"Total removed: "
        f"{input_rows - rows_after_cleaning:,}"
    )

    print(
        f"Saved: {final_output}"
    )

print("\n" + "=" * 90)
print("        FINAL PREPROCESSING COMPLETE")
print("=" * 90)
```

```python
print("=" * 90)
print("        FINAL CIC-IDS2017 DATASET INTEGRITY CHECK")
print("=" * 90)

final_files = sorted(
    glob.glob(
        os.path.join(
            FINAL_PROCESSED_DIR,
            "final_*.csv"
        )
    )
)

final_integrity = []

for file in final_files:

    file_name = os.path.basename(file)

    total_rows = 0
    total_missing = 0
    total_infinite = 0
    total_duplicates = 0
    total_missing_labels = 0

    all_labels = set()

    for chunk in pd.read_csv(
        file,
        chunksize=CHUNK_SIZE,
        low_memory=False
    ):

        total_rows += len(chunk)

        feature_cols = [
            column
            for column in chunk.columns
            if column != "Label"
        ]

        total_missing += int(
            chunk[feature_cols]
            .isna()
            .sum()
            .sum()
        )

        total_infinite += int(
            np.isinf(
                chunk[feature_cols]
            ).sum()
            .sum()
        )

        total_duplicates += int(
            chunk.duplicated().sum()
        )

        total_missing_labels += int(
            chunk["Label"].isna().sum()
        )

        all_labels.update(
            chunk["Label"]
            .dropna()
            .unique()
        )

    final_integrity.append({
        "File": file_name,
        "Rows": total_rows,
        "Columns": len(feature_cols) + 1,
        "Missing": total_missing,
        "Infinite": total_infinite,
        "Duplicates": total_duplicates,
        "Missing_Labels": total_missing_labels,
        "Classes": len(all_labels)
    })

final_integrity_df = pd.DataFrame(
    final_integrity
)

display(final_integrity_df)

print("\n" + "=" * 90)

print(
    "Total final rows:",
    f"{final_integrity_df['Rows'].sum():,}"
)

print(
    "Total missing values:",
    f"{final_integrity_df['Missing'].sum():,}"
)

print(
    "Total infinite values:",
    f"{final_integrity_df['Infinite'].sum():,}"
)

print(
    "Total duplicate rows:",
    f"{final_integrity_df['Duplicates'].sum():,}"
)

print(
    "Total missing labels:",
    f"{final_integrity_df['Missing_Labels'].sum():,}"
)

print(
    "Number of final files:",
    len(final_files)
)

print("\n" + "=" * 90)

if (
    final_integrity_df["Columns"].eq(79).all()
    and final_integrity_df["Missing"].sum() == 0
    and final_integrity_df["Infinite"].sum() == 0
    and final_integrity_df["Duplicates"].sum() == 0
    and final_integrity_df["Missing_Labels"].sum() == 0
):
    print("✅ FINAL DATASET PASSED INTEGRITY CHECK")
else:
    print("⚠️ FINAL DATASET REQUIRES INVESTIGATION")
```

```python
print("=" * 90)
print("        FINAL CIC-IDS2017 CLASS DISTRIBUTION")
print("=" * 90)

final_label_counts = {}

for file in final_files:

    file_name = os.path.basename(file)

    for chunk in pd.read_csv(
        file,
        chunksize=CHUNK_SIZE,
        usecols=["Label"],
        low_memory=False
    ):

        counts = chunk["Label"].value_counts()

        for label, count in counts.items():
            final_label_counts[label] = (
                final_label_counts.get(label, 0)
                + int(count)
            )

final_class_distribution = pd.DataFrame(
    list(final_label_counts.items()),
    columns=["Label", "Count"]
)

final_class_distribution = (
    final_class_distribution
    .sort_values(
        "Count",
        ascending=False
    )
    .reset_index(drop=True)
)

final_class_distribution["Percentage"] = (
    final_class_distribution["Count"]
    / final_class_distribution["Count"].sum()
    * 100
)

print("\nFinal class distribution:")
display(final_class_distribution)

print("\n" + "=" * 90)

print(
    "Total rows counted:",
    f"{final_class_distribution['Count'].sum():,}"
)

print(
    "Number of classes:",
    len(final_class_distribution)
)
```
